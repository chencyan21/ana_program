```python
import subprocess
import os
import shutil

def compress_pdf_ghostscript(input_path: str, output_path: str, quality: str = 'ebook'):
    """
    使用Ghostscript压缩PDF文件。

    Args:
        input_path (str): 输入PDF文件的路径。
        output_path (str): 输出压缩后PDF文件的路径。
        quality (str): 压缩质量预设。
                       可选值: 'screen', 'ebook', 'printer', 'prepress', 'default'。
    
    Returns:
        bool: 如果成功则返回True，否则返回False。
    """
    # 检查Ghostscript命令是否存在
    gs_command = shutil.which('gs') or shutil.which('gswin64c') or shutil.which('gswin32c')
    if not gs_command:
        print("错误：未找到Ghostscript。请确保已安装并将其添加至系统PATH。")
        return False

    print(f"正在使用Ghostscript压缩文件: {input_path}")
    print(f"压缩质量设置为: {quality}")

    try:
        # 构建Ghostscript命令
        command = [
            gs_command,
            '-sDEVICE=pdfwrite',
            '-dCompatibilityLevel=1.4',
            f'-dPDFSETTINGS=/{quality}',
            '-dNOPAUSE',
            '-dQUIET',
            '-dBATCH',
            f'-sOutputFile={output_path}',
            input_path
        ]

        # 执行命令
        subprocess.run(command, check=True, capture_output=True, text=True)

        # 比较文件大小
        original_size = os.path.getsize(input_path) / (1024 * 1024)  # MB
        compressed_size = os.path.getsize(output_path) / (1024 * 1024)  # MB
        
        print("压缩成功！")
        print(f"  - 原始文件大小: {original_size:.2f} MB")
        print(f"  - 压缩后文件大小: {compressed_size:.2f} MB")
        print(f"  - 压缩率: {(1 - compressed_size / original_size) * 100:.2f}%")
        
        return True

    except subprocess.CalledProcessError as e:
        print("Ghostscript执行失败。")
        print(f"错误信息: {e.stderr}")
        return False
    except FileNotFoundError:
        print(f"错误: 输入文件未找到{input_path}")
        return False

# --- 使用示例 ---
if __name__ == '__main__':
    # 自动检测当前目录下的PDF文件
    pdf_files = [f for f in os.listdir('.') if f.lower().endswith('.pdf')]
    if pdf_files:
        input_file = pdf_files[0]  # 选择第一个PDF文件
        print(f"检测到PDF文件: {input_file}")
    else:
        input_file = ''
        print("当前目录下未检测到PDF文件。")

    
    # 假设你有一个名为 'large_document.pdf' 的文件在同一目录下
    if os.path.exists(input_file):
        # 使用'ebook'设置进行高质量压缩，以平衡大小和质量
        # quality_options=['screen', 'ebook', 'printer', 'prepress', 'default']
        # for quality in quality_options:
        quality = 'printer'
        print(f"正在使用Ghostscript进行{quality}压缩...")
        output_file = f'C:/Users/C1602/Desktop/{input_file}.pdf'
        if compress_pdf_ghostscript(input_file, output_file, quality):
            print(f"压缩完成: {output_file}")
        else:
            print(f"压缩失败: {output_file}")
        # compress_pdf_ghostscript(input_file, output_file, quality='ebook')
    else:
        print(f"示例文件'{input_file}'不存在，请替换为您的文件路径。")

```
